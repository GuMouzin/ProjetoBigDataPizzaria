```python
teste = pd.read_csv('teste.csv')
```

```python
sobre_pizzas = teste[['pizza_id', 'pizza_name_id', 'pizza_category', 'pizza_ingredients', 'pizza_name']]
```

```python
sobre_pizzas['pizza_name_id'] = sobre_pizzas['pizza_name_id'].str.replace('_l|_m|_s|_xl', '', regex=True)
```

```python
import random
```

```python
unit_prices = {}

for i, prices in sobre_pizzas.iterrows():
    unit_prices[prices['pizza_name_id']] = round(random.uniform(16.00, 25.00), 2)

unit_prices

# hawaiian_s = unit_prices['hawaiian']
# hawaiian_m = unit_prices['hawaiian'] * 1.05
# hawaiian_l = unit_prices['hawaiian'] * 1.10
```

```python
dados = {
    'pizza_id':          [0]   * len(sobre_pizzas) * 4,
    'pizza_name_id':     ['']  * len(sobre_pizzas) * 4,
    'pizza_size':        ['']  * len(sobre_pizzas) * 4,
    'pizza_category':    ['']  * len(sobre_pizzas) * 4,
    'pizza_ingredients': ['']  * len(sobre_pizzas) * 4,
    'pizza_name':        ['']  * len(sobre_pizzas) * 4,
    'unit_price':        [0.0] * len(sobre_pizzas) * 4,
}

df = pd.DataFrame(dados)

# df.at[0, 'pizza_id'] = 0

#df
```

```python
idx = 0
for i, prices in sobre_pizzas.iterrows():
    for t in ['_s', '_m', '_l', '_xl']:
        #print(idx, df.pizza_id.dtype)
        df.at[idx, 'pizza_id'] = idx
        df.at[idx, 'pizza_name_id'] = prices['pizza_name_id'] + t
        df.at[idx, 'pizza_category'] = prices['pizza_category']
        df.at[idx, 'pizza_ingredients'] = prices['pizza_ingredients']
        df.at[idx, 'pizza_name'] = prices['pizza_name']
        match t:
            case '_s':
                df.at[idx, 'pizza_size'] = 'S'
                df.at[idx, 'unit_price'] = round(unit_prices[prices['pizza_name_id']], 2)
            case '_m':
                df.at[idx, 'pizza_size'] = 'M'
                df.at[idx, 'unit_price'] = round(unit_prices[prices['pizza_name_id']] * 1.05, 2)
            case '_l':
                df.at[idx, 'pizza_size'] = 'L'
                df.at[idx, 'unit_price'] = round(unit_prices[prices['pizza_name_id']] * 1.10, 2)
            case '_xl':
                df.at[idx, 'pizza_size'] = 'XL'
                df.at[idx, 'unit_price'] = round(unit_prices[prices['pizza_name_id']] * 1.15, 2)
            case _:
                pass
        idx += 1

#df.pizza_id.astype('int')
# df

# hawaiian_s = unit_prices['hawaiian']
# hawaiian_m = unit_prices['hawaiian'] * * 1.10
```
